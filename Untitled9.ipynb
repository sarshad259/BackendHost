```python
%%writefile data.yaml
path: .
train: images/train
val: images/valid
test: images/test

names:
  0: no_mask
  1: mask
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data="data.yaml",
    epochs=30,
    imgsz=640,
    batch=16
)
```

```python
model.predict(
    source="images/test",
    conf=0.25,
    save=True
)
```
